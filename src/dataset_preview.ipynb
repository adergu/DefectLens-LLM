```python
# dataset_preview.ipynb
# Generic Dataset Preview Notebook

import pandas as pd
from IPython.display import display, HTML

# --- FIX: show all columns ---
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 40)

# --- CONFIG: change this to your dataset ---
#FILE = "../data/04_llm_scores/django_2020_2025_12D_LLM_FINAL.parquet"
#FILE = "../data/04_llm_scores/fastapi_2020_2025_12D_LLM_FINAL.parquet"
#FILE = "../data/04_llm_temp/fastapi_2020_2025_12D_LLM_FINAL.parquet"
#FILE = "../data/04_llm_scores/pandas_2020_2025_12D_LLM_FINAL.parquet"
#FILE = "../data/04_llm_scores/numpy_2020_2025_12D_LLM_FINAL.parquet"
#FILE = "../data/02_snapshots/django_2020_2025_bimonthly_CLEAN.parquet"
#FILE = "../data/02_snapshots/fastapi_2020_2025_bimonthly_CLEAN.parquet"
#FILE = "../data/02_snapshots/pandas_2020_2025_bimonthly_CLEAN.parquet"
#FILE = "../data/03_labeled/pandas_2020_2025_FINAL_LABELED.parquet"
#FILE = "../data/03_labeled/fastapi_2020_2025_FINAL_LABELED_TOP50.parquet"
#FILE = "../data/03_labeled/fastapi_2020_2025_FINAL_LABELED.parquet"
#FILE = "../data/04_llm_temp/numpy_2020_2025_FINAL_LABELED_12D_LLM_GROQ.parquet"
FILE = "../data/04_llm_temp/fastapi_2020_2025_FINAL_LABELED_12D_LLM_GROQ.parquet"

# Load parquet
df = pd.read_parquet(FILE)

# --- exclude source_code column ---
df = df.drop(columns=["source_code"], errors="ignore")

# --- Tabular preview (first 20 rows) ---
# If 'file_path' exists, create a shortened version for readability
if "file_path" in df.columns:
    df['file_path_short'] = df['file_path'].apply(lambda x: "..." + x[-40:] if len(x) > 40 else x)

# Dynamically fetch all columns to display
cols_tabular = df.columns.tolist()

# --- FIX: well bordered scrollable HTML table ---
def show_table(dataframe, title="", n=50):
    html = f"""
    <h3 style="font-family: Arial; color: #1a1a2e; margin-bottom: 8px;">{title}</h3>
    <div style="
        overflow-x: auto;
        overflow-y: auto;
        max-height: 500px;
        max-width: 100%;
        border: 2px solid #1a1a2e;
        border-radius: 6px;
    ">
    <table style="
        border-collapse: collapse;
        width: 100%;
        font-family: Arial, sans-serif;
        font-size: 12px;
    ">
    <thead>
        <tr>
    """
    for col in dataframe.columns:
        html += f"""
            <th style="
                background-color: #1a1a2e;
                color: #c8a96e;
                padding: 8px 12px;
                text-align: left;
                border: 1px solid #444;
                white-space: nowrap;
                position: sticky;
                top: 0;
                z-index: 1;
            ">{col}</th>
        """
    html += "</tr></thead><tbody>"
    for i, row in dataframe.head(n).iterrows():
        bg = "#f9f9f9" if i % 2 == 0 else "#ffffff"
        html += f'<tr style="background-color: {bg};">'
        for col in dataframe.columns:
            val = row[col]
            # highlight is_defective
            if col == "is_defective":
                color = "#c0392b" if val == 1 else "#27ae60"
                html += f'<td style="padding: 6px 12px; border: 1px solid #ddd; white-space: nowrap; color: {color}; font-weight: bold;">{val}</td>'
            else:
                html += f'<td style="padding: 6px 12px; border: 1px solid #ddd; white-space: nowrap;">{val}</td>'
        html += "</tr>"
    html += "</tbody></table></div>"
    display(HTML(html))

print("### First 50 rows (all columns, scrollable)")
show_table(df[cols_tabular], title="Dataset Preview — First 50 Rows", n=50)

# --- Card-style preview (first 10 rows) ---
print("\n### First n rows (card-style)")
for i, row in df.head(10).iterrows():
    print("="*80)
    for col in cols_tabular:
        print(f"{col:<25}: {row[col]}")

# --- Columns and metadata ---
print("\n### Columns in dataset")
print(cols_tabular)

# --- If repository_name exists, print unique values ---
if "repository_name" in df.columns:
    print("\n### Unique repository_name values")
    print(df["repository_name"].unique())
```
